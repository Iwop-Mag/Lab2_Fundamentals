```python
# [CELL 1] CONFIGURATION AND PASSWORD GENERATION
ENGINEER = "Magalong"
STUDENT_ID = "TUPM-26-1134"

SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")

secretCredentials = '@' + ENGINEER.strip().upper() + '_' + str(SEED_NUM ** 3)
print(f'\nPassword for the Account: {secretCredentials}')
```

```python
# [CELL 2] PASSWORD GENERATION

def authentication():
    for i in range(0, SEED_NUM + 1):
        entered = input('Enter Password ...')
        if entered == secretCredentials:
            print(f'{i} incorrect attempts.\nACCESS GRANTED.')
            return
    print(f'Maximum incorrect attempts of {i + 1} reached.\nACCESS DENIED.')

authentication()
```
