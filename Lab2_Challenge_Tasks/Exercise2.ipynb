```python
# [CELL 1] CONFIGURATION
ENGINEER = "Magalong"
STUDENT_ID = "TUPM-26-1134"

SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] SIGNAL GENERATION
import statistics

def multiplyBySeed(x):
    return SEED_NUM * x

signalText = f"{ENGINEER} {SEED_NUM}"   # the text signal to analyze in Cell 5
signalData = [multiplyBySeed(ord(c)) for c in ENGINEER]
meanData = statistics.mean(signalData)  # boundary, ±5% tolerance

print(f"### Received Signals from {ENGINEER} ###\n")
for i, value in enumerate(signalData, start=1):
    print(f"Signal Content {i}: {value}")
print(f"\nAverage: {meanData}")
```

```python
# [CELL 3] DATA VALIDATION AND FILTERING
def validateData(signalData, meanData):
    lower = meanData * 0.95
    upper = meanData * 1.05
    print(f"### Acceptable Signal Range: {lower:.2f} - {upper:.2f} ###\n")

    for i, value in enumerate(signalData, start=1):
        if value < lower:
            print(f"Signal Content {i}: {value} --> too low")
        elif value > upper:
            print(f"Signal Content {i}: {value} --> too high")
        else:
            print(f"Signal Content {i}: {value} --> accepted")

    return [v for v in signalData if lower <= v <= upper]

validSignal = validateData(signalData, meanData)
print(f"\nAccepted Data: {validSignal}")
```

```python
# [CELL 4] SIGNAL PROCESSING AND NORMALIZATION
minData = min(validSignal)
maxData = max(validSignal)
rng = maxData - minData

normData = [(x - minData) / rng for x in validSignal] if rng != 0 else [0.0] * len(validSignal)
print(f"Normalized Signal: {normData}")
```

```python
# [CELL 5] CHARACTER-BY-CHARACTER ANALYSIS AND CLASSIFICATION
letters = digits = spaces = others = 0

print(f"### Analyzing Signal: '{signalText}' ###\n")
for i, ch in enumerate(signalText, start=1):
    if ch.isalpha():
        kind = "letter"
        letters += 1
    elif ch.isdigit():
        kind = "digit"
        digits += 1
    elif ch.isspace():
        kind = "space"
        spaces += 1
    else:
        kind = "other"
        others += 1
    print(f"Character {i}: '{ch}' --> {kind}")

print(f'\n=================================================\n')

total = len(signalText)
print(f"### DIAGNOSTIC REPORT ###\n")
print(f"Total characters : {total}")
print(f"Letters          : {letters} ({letters / total:.0%})")
print(f"Digits           : {digits} ({digits / total:.0%})")
print(f"Spaces           : {spaces} ({spaces / total:.0%})")
print(f"Other symbols    : {others} ({others / total:.0%})")
print(f"Accepted readings: {len(validSignal)} of {len(signalData)}")

if letters > total / 2:
    classification = "TEXT-DOMINANT SIGNAL"
elif digits > total / 2:
    classification = "NUMERIC-DOMINANT SIGNAL"
else:
    classification = "MIXED SIGNAL"
print(f"\nFinal Classification: {classification}")
```
