```python
# [CELL 1] CONFIGURATION
ENGINEER = "Magalong"
STUDENT_ID = "TUPM-26-1134"

SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] GENERATION OF UNIQUE TELEMETRY/SENSOR DATA


import statistics

def multiplyBySeed(x): # to avoid repitative tasks sir pogi
    x = SEED_NUM * x
    return(x)

telemetryEntries = len(ENGINEER)

sensorData = [multiplyBySeed(ord(x)) for x in ENGINEER]
meanTelemetry = statistics.mean(sensorData) # setting this as the boundary, adding a ±5% tolerance for accepted readings


print(f"### Received Sensor Readings from {ENGINEER} ###")
for x in range(1, telemetryEntries + 1):
    print(f'Sensor Data {x}: {sensorData[x - 1]}')
print("Average: ", meanTelemetry)
```

```python
# [CELL 3] DATA VALIDATION AND FILTERING

def validateSensorData(sensorData, meanTelemetry):
    lower = meanTelemetry * 0.95
    upper = meanTelemetry * 1.05

    for i, value in enumerate(sensorData, start=1):
        if value < lower:
            print(f'Sensor Data {i}: {value} --> too low')
        elif value > upper:
            print(f'Sensor Data {i}: {value} --> too high')
        else:
            print(f'Sensor Data {i}: {value} --> accepted')

    validData = [v for v in sensorData if lower <= v <= upper]
    return validData


validData = validateSensorData(sensorData, meanTelemetry)
print('Accepted Data:', validData)
```
